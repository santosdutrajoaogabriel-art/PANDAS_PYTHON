Sua tarefa é realizar alguns tratamentos e deixar esse arquivo JSON como um Pandas DataFrame, sem dados vazios e com a tipagem das colunas de forma correta. Como isso pode ser feito?

```python
import pandas as pd
```

```python
pd.read_json('/content/dados.json')
```

```python
import json
```

```python
with open('dados.json', 'r') as f:
  json_main = json.load(f)
```

```python
json_main
```

```python
df = pd.json_normalize(json_main, record_path='pessoas')
```

```python
df
```

- identificando os tipos de colunas e se possui valores nulos.

```python
df.info()
```

```python
df['idade'] = df['idade'].astype(int)
```
